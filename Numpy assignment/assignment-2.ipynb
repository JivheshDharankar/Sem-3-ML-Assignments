Q9. Award medals to the scores below using nested np.where: 'Gold' for 90 and above, 'Silver' for 75
and above, 'Bronze' for 60 and above, and 'None' otherwise. Then count how many students received
each medal.
scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

```python
scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])

medals = np.where(scores >= 90, 'Gold',
         np.where(scores >= 75, 'Silver',
         np.where(scores >= 60, 'Bronze', 'None')))

labels, counts = np.unique(medals, return_counts=True)

print(medals)
print(dict(zip(labels, counts)))
```

Q14. Build a complete multi-step pipeline. Starting from the two arrays below: convert them into column
vectors, join them into a single two-column table, attach a third column holding each row's total, then
multiply the three-column table by w = [0.5, 0.2, 0.3] to obtain one score per row. Finally use
np.where to mark each score as 1 if it is above the average score and 0 otherwise. Report the shape of
every intermediate array.
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

```python


hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])


c1 = hours[:, np.newaxis]
c2 = attendance[:, np.newaxis]
print("Column 1 shape:", c1.shape)
print("Column 2 shape:", c2.shape)


table = np.hstack((c1, c2))
print("Table shape:", table.shape)


totals = table.sum(axis=1, keepdims=True)
table_3col = np.hstack((table, totals))
print("3-column table shape:", table_3col.shape)


w = np.array([0.5, 0.2, 0.3])
scores = table_3col @ w
print("Scores shape:", scores.shape)


final_labels = np.where(scores > scores.mean(), 1, 0)
print("Final labels shape:", final_labels.shape)
print("Final labels:", final_labels)
```

Q15. Apply min-max normalization to scale any array into the 0 to 1 range using (x - min) / (max -

Q20. Start from arr = np.arange(1, 11). Create one variable using a plain slice arr[2:6] and
another using arr[2:6].copy(). Change the first element of each to 999, then print the original array
and explain in a comment which change survived and why.

```python
arr=np.arange(1,11)
var1 = arr[2:6]
var2 = arr[2:6].copy()
var1[0] = 999
var2[0] = 999
print(arr)
print(var1)
print(var2)

# Only the change made through the plain slice (var1[0] = 999) survived in the original array.
# 
# Why? 
# 1. In NumPy, a plain slice creates a 'view' of the data. It points back to the exact same 
#    memory block as the original array. Changing the view directly overwrites the parent array.
# 2. Using .copy() allocates entirely new, independent memory for var2. Modifications made 
#    there remain isolated and have zero effect on the original array.
```
