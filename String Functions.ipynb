```python
def check_sku(sku):
    cleaned = sku.strip().upper()
    if len(cleaned) != 8:
        return "Invalid SKU"
    letters = cleaned[0:3]
    hyphen = cleaned[3]
    digits = cleaned[4:8]
    if letters.isalpha() and hyphen == "-" and digits.isdigit():
        return f"Valid: {cleaned}"
    else:
        return "Invalid SKU"
```

```python
def valid(email):
    if "@" not in email:
        return "Invalid"
    parts = email.split("@")
    domain_part = parts[1]
    domain = domain_part.split(".")[0]
    return domain.title()
```

Question 8 · The Domain Name Slicer

```python
print(extensions("report.pdf, image.JPG, data.csv, notes.txt, photo.jpg"))
```

```python
def extensions(filenames):
    files = filenames.split(", ")
    exts = [file.split(".")[-1].upper() for file in files]
    unique_exts = list(dict.fromkeys(exts))
    return unique_exts
```

Question 6 · The Hashtag Extractor

```python
print(ispal("Was it a car or a cat I saw?"))
```

```python
def ispal(sentence):
    cleaned = "".join(ch for ch in sentence if ch.isalnum()).lower()
    rev = cleaned[::-1]
    return rev == cleaned
```

Question 3 · The Messy Database Cleaner

```python
print(reverser("charlie-bravo-alpha"))
```

```python
def reverser(code):
    cleaned_code = code.replace("-"," ")
    words = cleaned_code.split()
    reversed_words = words[::-1]
    ans = " ".join(reversed_words)
    return ans.upper()
```
