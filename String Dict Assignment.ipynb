```python
movies = [
 {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
 {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
 {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
 {"title": "Movie D", "rating": 8.2, "genre": "Drama"},
]
movie_analysis(movies)
```

```python
def movie_analysis(movies):
    high_rated = [m["title"] for m in movies if m["rating"] > 8.1]
    print("Movies with rating > 8.1:")
    for t in high_rated:
        print(f"- {t}")

    genres = list(dict.fromkeys(m["genre"] for m in movies))
    print(f"Genres: {genres}")

    grouped = {}
    for m in movies:
        grouped.setdefault(m["rating"], []).append(m["title"])
    print("Movies grouped by rating:")
    for rating in sorted(grouped.keys(), reverse=True):
        print(f"{rating}: {grouped[rating]}")
```

Question 9 · The Salary Sorter

```python
test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}
print(merge_marks(test1, test2))
```

```python
def merge_marks(test1, test2):
    merged = dict(test1)
    for k, v in test2.items():
        merged[k] = merged.get(k, 0) + v
    return merged
```

Question 8 · The Marks Merger

```python
sales = {"mon": 200, "tue": 350, "wed": 200}
print(invert_dict(sales))
```

```python
def invert_dict(d):
    inverted = {}
    for k, v in d.items():
        if v in inverted:
            inverted[v].append(k)
        else:
            inverted[v] = [k]
    return inverted
```

Question 7 · The Dictionary Inverter

```python
word_frequency("the cat sat on the mat the cat ran")
```

```python
def word_frequency(sentence):
    words = sentence.lower().split()
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    most_common = sorted(freq.items(), key=lambda x: x[1], reverse=True)[0]
    print(freq)
    print(f"Most frequent word: {most_common[0]} ({most_common[1]} times)")
```

Question 6 · The Word Frequency Counter

```python
print(char_frequency("Aba c"))
```

```python
def char_frequency(s):
    cleaned = s.replace(" ", "").lower()
    freq = {}
    for ch in cleaned:
        freq[ch] = freq.get(ch, 0) + 1
    return freq
```

Question 5 · The Character Frequency Counter

```python
sentence_stats("Python is an amazing programming language for data")
```

```python
def sentence_stats(sentence):
    words = sentence.split()
    total_words = len(words)
    vowels = "aeiou"
    vowel_end_count = sum(1 for w in words if w[-1].lower() in vowels)
    longest_word = max(words, key=len)
    print(f"Total words : {total_words}")
    print(f"Words ending with vowel : {vowel_end_count}")
    print(f"Longest word : {longest_word}")
```

Question 4 · The Sentence Analyzer

```python
print(validate_username2("Rahul_123"))
print(validate_username2("rahul123"))
```

```python
def validate_username2(username):
    cleaned = username.strip().lower()
    if cleaned.isalnum() and cleaned[0].isalpha() and 5 <= len(cleaned) <= 12:
        return "Valid username"
    else:
        return "Invalid username"
```

Question 3 · The Username Validator

```python
price_stats(" 120, 45 , 300 ,80, 95 ")
```

```python
def price_stats(prices_str):
    prices = [int(p.strip()) for p in prices_str.split(",")]
    total = sum(prices)
    highest = max(prices)
    average = total / len(prices)
    print(f"Total : {total}")
    print(f"Highest : {highest}")
    print(f"Average : {average:.2f}")
```

Question 2 · The Messy Price List Calculator

```python
print(clean_city(" nEw dELhi "))
```

```python
def clean_city(city):
    cleaned = city.strip().title()
    if all(ch.isalpha() or ch == ' ' for ch in cleaned):
        return cleaned
    else:
        return "Invalid city"
```
