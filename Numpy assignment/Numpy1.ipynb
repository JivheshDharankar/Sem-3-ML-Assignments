Q6. Given a = np.array([2, 4, 6]) and b = np.array([2.0, 4.0, 6.0]) - check both dtypes, then convert b
to int with astype and verify the dtypes now match.

```python
a = np.array([2,4,6])
b = np.array([2.0, 4,0, 6,0])
print(a.dtype)
print(b.dtype)
c = b.astype(int)
print(a.dtype)
print(c.dtype)
```
