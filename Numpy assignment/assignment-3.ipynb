Q15 — The final weighted result

```python
M = np.array([[70, 65, 80],
              [55, 90, 72],
              [88, 76, 91],
              [60, 58, 66]])
w = np.array([0.3, 0.3, 0.4])
W = np.array([[0.3, 0.5],
              [0.3, 0.2],
              [0.4, 0.3]])
 
final15 = M @ w                                                                                                                # 1
print(np.round(final15, 2))
print(final15.argmax())                                                                                                          # 2
 
subject_totals = np.ones(M.shape[0]) @ M                                                                                            # 3 -> matches expected
print(subject_totals)
 
final2 = M @ W                                                                                                                         # 4 -> matches expected
print(np.round(final2, 2))
 
print(np.sum(final2[:, 1] > final2[:, 0]))                                                                                                # 5
 
diff15 = np.round(final2[:, 1] - final2[:, 0], 2)                                                                                           # 6 -> matches expected
print(diff15)
 
better = np.max(final2, axis=1)
result15 = np.hstack((M, better.reshape(-1, 1)))                                                                                              # 7
print(result15)
```
