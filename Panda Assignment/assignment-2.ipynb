B4.
Out of all the passengers in first class, what percentage survived? Now do the same for third class. Put the two
numbers next to each other in a comment.

```python
class1_total = t[t['Pclass'] == 1].shape[0]
survived = t[(t['Pclass'] == 1) & (t['Survived'] == 1)].shape[0]
ans = (survived / class1_total) * 100

class3_total = t[t['Pclass'] == 3].shape[0]
survived3 = t[(t['Pclass'] == 3) & (t['Survived'] == 1)].shape[0]
ans2 = (survived3 / class3_total) * 100
print(ans)
print(ans2)
```
